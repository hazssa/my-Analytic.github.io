```python
import pandas as pd
import matplotlib.pyplot as plt
import textwrap

# =========================
# 1. Prepare the data
# =========================

# Make sure Total_Sales is numeric
df["Total_Sales"] = pd.to_numeric(df["Total_Sales"], errors="coerce")

# Remove rows where Total_Sales is missing
df = df.dropna(subset=["Total_Sales"])


# =========================
# 2. Create grouped data
# =========================

category_sales = (
    df.groupby("Category")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)

payment_sales = (
    df.groupby("Payment_Method")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)

top_customers = (
    df.groupby("Customer_Name")["Total_Sales"]
      .sum()
      .sort_values(ascending=True)
      .tail(10)
)

region_sales = (
    df.groupby("Region")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)


# =========================
# 3. Helper function for summaries
# =========================

def add_summary(ax, text):
    wrapped_text = textwrap.fill(text, width=65)
    ax.text(
        0.5, -0.28,
        wrapped_text,
        transform=ax.transAxes,
        ha="center",
        va="top",
        fontsize=10,
        fontweight="bold"
    )


# =========================
# 4. Create one image with 4 charts
# =========================

fig, axes = plt.subplots(2, 2, figsize=(18, 14))

fig.suptitle(
    "Sales Performance Dashboard",
    fontsize=22,
    fontweight="bold"
)

# =========================
# Chart 1: Sales by Category
# =========================

ax1 = axes[0, 0]

category_sales.plot(
    kind="bar",
    ax=ax1,
    title="Total Sales by Category"
)

ax1.set_xlabel("Category")
ax1.set_ylabel("Total Sales")
ax1.tick_params(axis="x", rotation=45)

for bar in ax1.patches:
    ax1.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{bar.get_height():,.0f}',
        ha="center",
        va="bottom",
        fontsize=9
    )

top_category = category_sales.idxmax()
top_category_value = category_sales.max()

summary1 = (
    f"{top_category} generated the highest sales with "
    f"{top_category_value:,.0f}, making it the strongest-performing category."
)

add_summary(ax1, summary1)


# =========================
# Chart 2: Sales by Payment Method
# =========================

ax2 = axes[0, 1]

payment_sales.plot(
    kind="pie",
    ax=ax2,
    autopct="%1.1f%%",
    startangle=90
)

ax2.set_title("Sales Distribution by Payment Method")
ax2.set_ylabel("")

top_payment = payment_sales.idxmax()
top_payment_value = payment_sales.max()
payment_total = payment_sales.sum()
top_payment_percent = (top_payment_value / payment_total) * 100

summary2 = (
    f"{top_payment} is the most used payment method, accounting for "
    f"{top_payment_percent:.1f}% of total sales."
)

add_summary(ax2, summary2)


# =========================
# Chart 3: Top 10 Customers by Sales
# =========================

ax3 = axes[1, 0]

top_customers.plot(
    kind="barh",
    ax=ax3,
    title="Top 10 Customers by Sales"
)

ax3.set_xlabel("Total Sales")
ax3.set_ylabel("Customer Name")

for bar in ax3.patches:
    ax3.text(
        bar.get_width(),
        bar.get_y() + bar.get_height() / 2,
        f'{bar.get_width():,.0f}',
        va="center",
        fontsize=9
    )

top_customer = top_customers.idxmax()
top_customer_value = top_customers.max()

summary3 = (
    f"{top_customer} is the highest-value customer among the top 10, "
    f"with total sales of {top_customer_value:,.0f}."
)

add_summary(ax3, summary3)


# =========================
# Chart 4: Sales by Region
# =========================

ax4 = axes[1, 1]

region_sales.plot(
    kind="bar",
    ax=ax4,
    title="Total Sales by Region"
)

ax4.set_xlabel("Region")
ax4.set_ylabel("Total Sales")
ax4.tick_params(axis="x", rotation=45)

for bar in ax4.patches:
    ax4.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{bar.get_height():,.0f}',
        ha="center",
        va="bottom",
        fontsize=9
    )

top_region = region_sales.idxmax()
top_region_value = region_sales.max()

summary4 = (
    f"{top_region} recorded the highest regional sales with "
    f"{top_region_value:,.0f}, making it the strongest sales region."
)

add_summary(ax4, summary4)


# =========================
# 5. Final layout and save image
# =========================

plt.subplots_adjust(
    hspace=0.55,
    wspace=0.25,
    top=0.90,
    bottom=0.08
)

plt.savefig(
    "sales_dashboard_summary.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor
from IPython.display import FileLink, display
import os

# -------------------------
# Clean Data
# -------------------------
df["Total_Sales"] = pd.to_numeric(df["Total_Sales"], errors="coerce")
df = df.dropna(subset=["Total_Sales"])

# -------------------------
# Create Grouped Data
# -------------------------
category_sales = (
    df.groupby("Category")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)

payment_sales = (
    df.groupby("Payment_Method")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)

top_customers = (
    df.groupby("Customer_Name")["Total_Sales"]
      .sum()
      .sort_values(ascending=True)
      .tail(10)
)

region_sales = (
    df.groupby("Region")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
)

# -------------------------
# Create Dashboard Image
# -------------------------
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
fig.suptitle(
    "Sales Performance Dashboard",
    fontsize=28,
    fontweight="bold"
)

# Chart 1: Category Sales
ax1 = axes[0, 0]
category_sales.plot(kind="bar", ax=ax1)
ax1.set_title("Total Sales by Category", fontsize=15, fontweight="bold")
ax1.set_xlabel("Category")
ax1.set_ylabel("Total Sales")
ax1.tick_params(axis="x", rotation=45)

for bar in ax1.patches:
    ax1.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha="center",
        va="bottom",
        fontsize=8
    )

# Chart 2: Payment Method
ax2 = axes[0, 1]
payment_sales.plot(
    kind="pie",
    ax=ax2,
    autopct="%1.1f%%",
    startangle=90
)
ax2.set_title("Sales Distribution by Payment Method", fontsize=15, fontweight="bold")
ax2.set_ylabel("")

# Chart 3: Top Customers
ax3 = axes[1, 0]
top_customers.plot(kind="barh", ax=ax3)
ax3.set_title("Top 10 Customers by Sales", fontsize=15, fontweight="bold")
ax3.set_xlabel("Total Sales")
ax3.set_ylabel("Customer Name")

for bar in ax3.patches:
    ax3.text(
        bar.get_width(),
        bar.get_y() + bar.get_height() / 2,
        f"{bar.get_width():,.0f}",
        va="center",
        fontsize=8
    )

# Chart 4: Region Sales
ax4 = axes[1, 1]
region_sales.plot(kind="bar", ax=ax4)
ax4.set_title("Total Sales by Region", fontsize=15, fontweight="bold")
ax4.set_xlabel("Region")
ax4.set_ylabel("Total Sales")
ax4.tick_params(axis="x", rotation=45)

for bar in ax4.patches:
    ax4.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{bar.get_height():,.0f}",
        ha="center",
        va="bottom",
        fontsize=8
    )

plt.tight_layout(rect=[0, 0, 1, 0.94])

dashboard_image = "sales_dashboard.png"
plt.savefig(dashboard_image, dpi=300, bbox_inches="tight")
plt.show()

# -------------------------
# Create Summary Insights
# -------------------------
summary_text = [
    f"{category_sales.idxmax()} generated the highest category sales with {category_sales.max():,.0f}.",
    f"{payment_sales.idxmax()} was the leading payment method with {payment_sales.max():,.0f} in sales.",
    f"{top_customers.idxmax()} was the highest-value customer with {top_customers.max():,.0f} in sales.",
    f"{region_sales.idxmax()} recorded the highest regional sales with {region_sales.max():,.0f}."
]

# -------------------------
# Create PowerPoint
# -------------------------
prs = Presentation()
prs.slide_width = Inches(13.33)
prs.slide_height = Inches(7.5)

# Theme Colors
dark_blue = RGBColor(14, 35, 64)
gold = RGBColor(230, 174, 60)
light_bg = RGBColor(245, 247, 250)
white = RGBColor(255, 255, 255)
grey = RGBColor(90, 90, 90)

# -------------------------
# Slide 1: Title Slide
# -------------------------
slide = prs.slides.add_slide(prs.slide_layouts[6])

bg = slide.background.fill
bg.solid()
bg.fore_color.rgb = dark_blue

title = slide.shapes.add_textbox(Inches(0.8), Inches(2.0), Inches(12), Inches(1))
tf = title.text_frame
tf.text = "Sales Performance Dashboard"
tf.paragraphs[0].font.size = Pt(44)
tf.paragraphs[0].font.bold = True
tf.paragraphs[0].font.color.rgb = white

subtitle = slide.shapes.add_textbox(Inches(0.85), Inches(3.2), Inches(11), Inches(0.7))
tf = subtitle.text_frame
tf.text = "Category • Payment Method • Customer • Regional Sales Analysis"
tf.paragraphs[0].font.size = Pt(20)
tf.paragraphs[0].font.color.rgb = gold

footer = slide.shapes.add_textbox(Inches(0.85), Inches(6.55), Inches(11), Inches(0.5))
tf = footer.text_frame
tf.text = "Built by Muhammed Hassan | hazssanmuh@gmail.com"
tf.paragraphs[0].font.size = Pt(14)
tf.paragraphs[0].font.color.rgb = white

# -------------------------
# Slide 2: Dashboard Image
# -------------------------
slide = prs.slides.add_slide(prs.slide_layouts[6])

bg = slide.background.fill
bg.solid()
bg.fore_color.rgb = white

title = slide.shapes.add_textbox(Inches(0.5), Inches(0.25), Inches(12), Inches(0.5))
tf = title.text_frame
tf.text = "Combined Sales Dashboard"
tf.paragraphs[0].font.size = Pt(30)
tf.paragraphs[0].font.bold = True
tf.paragraphs[0].font.color.rgb = dark_blue

slide.shapes.add_picture(
    dashboard_image,
    Inches(0.45),
    Inches(0.85),
    width=Inches(12.4)
)

footer = slide.shapes.add_textbox(Inches(0.5), Inches(7.05), Inches(12), Inches(0.3))
tf = footer.text_frame
tf.text = "Built by Muhammed Hassan | hazssanmuh@gmail.com"
tf.paragraphs[0].font.size = Pt(10)
tf.paragraphs[0].font.color.rgb = grey

# -------------------------
# Slide 3: Summary Insights
# -------------------------
slide = prs.slides.add_slide(prs.slide_layouts[6])

bg = slide.background.fill
bg.solid()
bg.fore_color.rgb = light_bg

title = slide.shapes.add_textbox(Inches(0.7), Inches(0.45), Inches(12), Inches(0.7))
tf = title.text_frame
tf.text = "Key Sales Insights"
tf.paragraphs[0].font.size = Pt(36)
tf.paragraphs[0].font.bold = True
tf.paragraphs[0].font.color.rgb = dark_blue

y_position = 1.5

for insight in summary_text:
    box = slide.shapes.add_textbox(Inches(1), Inches(y_position), Inches(11.3), Inches(0.75))
    tf = box.text_frame
    tf.text = "• " + insight
    tf.paragraphs[0].font.size = Pt(22)
    tf.paragraphs[0].font.color.rgb = RGBColor(40, 40, 40)
    y_position += 1.1

footer = slide.shapes.add_textbox(Inches(0.8), Inches(6.85), Inches(12), Inches(0.4))
tf = footer.text_frame
tf.text = "Built by Muhammed Hassan | hazssanmuh@gmail.com"
tf.paragraphs[0].font.size = Pt(12)
tf.paragraphs[0].font.color.rgb = grey

# -------------------------
# Save PowerPoint
# -------------------------
pptx_file = "Sales_Performance_Dashboard_Muhammed_Hassan.pptx"
prs.save(pptx_file)

print("PowerPoint created successfully!")
print("File saved as:", os.path.abspath(pptx_file))

# Clickable download link inside Jupyter Notebook
display(FileLink(pptx_file))
```
